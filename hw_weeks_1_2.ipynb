### Exercice 4 – Somme de deux sinusoïdes
Soit $x(t) = 3\cos(2\pi t) + 4\sin(2\pi t)$.

1. Mettre $x(t)$ sous la forme $A\cos(2\pi t - \phi)$ avec $A>0$ et $\phi\in[0,2\pi[$.  
2. Vérifier en traçant les deux signaux et leur somme sur l’intervalle $[0,1]$ s.  
3. Généraliser : pour $a\cos(\omega t) + b\sin(\omega t)$, donner les formules de $A$ et $\phi$.
### reponse Exercice 4 
On cherche $A\cos(\omega t-\phi)=A\cos\phi\cos\omega t+A\sin\phi\sin\omega t$.
Par identification : $A\cos\phi=3$, $A\sin\phi=4$ donc
$A=\sqrt{3^2+4^2}=5$, $\phi=\operatorname{atan2}(4,3)\approx 0{,}927$ rad ($\approx 53{,}13^\circ$).

**Généralisation :** pour $a\cos\omega t+b\sin\omega t$ :
$$A=\sqrt{a^2+b^2},\qquad \phi=\operatorname{atan2}(b,a)\ (\bmod 2\pi)$$

**Réponse** (à compléter) :

```python
# Vérification graphique
import numpy as np
import matplotlib.pyplot as plt
a, b = 3, 4
t = np.linspace(0, 1, 1000)
x = 3*np.cos(2*np.pi*t) + 4*np.sin(2*np.pi*t)

# Déterminer A et phi (à calculer)
A = np.hypot(a, b)
phi =  np.arctan2(b, a) % (2*np.pi)  # car cos(phi) = a/A, sin(phi) = b/A
x_phaseur = A*np.cos(2*np.pi*t - phi)
print(f"A = {A}, φ = {phi:.4f} rad = {np.degrees(phi):.2f}°")
plt.plot(t, a*np.cos(2*np.pi*t), ':', label='3cos(2πt)')
plt.plot(t, b*np.sin(2*np.pi*t), ':', label='4sin(2πt)')

plt.plot(t, x, label='somme')
plt.plot(t, x_phaseur, '--', label='A cos(2πt - φ)')
plt.legend()
plt.grid()
plt.show()
plt.xlabel('t (s)')
```

### Exercice 6 – Battements
Deux signaux de fréquences proches :  
$
x_1(t) = \cos(2\pi f_1 t)
$ et 
$
x_2(t) = \cos(2\pi f_2 t)
$.  
Montrer que leur somme peut s’écrire  
$$ 
2\cos(2\pi f_m t)\cos(2\pi f_d t) 
$$  
avec $f_m = \frac{f_1+f_2}{2}$ et $f_d = \frac{f_1-f_2}{2}$. 

$\cos p+\cos q=2\cos\frac{p+q}{2}\cos\frac{p-q}{2}$, pour $p=2\pi f_1t$, $q=2\pi f_2t$ :
$$x_1+x_2=2\cos(2\pi f_mt)\cos(2\pi f_dt),\quad f_m=\tfrac{f_1+f_2}{2},\ f_d=\tfrac{f_1-f_2}{2}.$$


Pour $f_1 = 10$ Hz et $f_2 = 12$ Hz, tracer la somme et observer l’enveloppe. Expliquer le phénomène de battement.

Pour $f_1=10$, $f_2=12$ : $f_m=11$ Hz (oscillation rapide), $f_d=-1$ Hz (cosinus pair, donc enveloppe à 1 Hz).
L'amplitude $|2\cos(2\pi f_dt)|$ s'annule 2 fois par seconde : on entend/voit **2 battements par seconde**, soit $|f_1-f_2|=2$ Hz.
Physiquement, les deux sinusoïdes passent alternativement en phase (interférence constructive, amplitude 2) et en opposition (destructive, amplitude 0).
**Démonstration** (à compléter) :

```python
f1 = 10
f2 = 12
t = np.linspace(0, 1, 2000)

x1 = np.cos(2*np.pi*f1*t)
x2 = np.cos(2*np.pi*f2*t)
x_sum = x1 + x2

fm = (f1+f2)/2
fd = (f1-f2)/2
x_env = 2*np.cos(2*np.pi*fd*t)  # enveloppe
x_product = x_env * np.cos(2*np.pi*fm*t)

plt.plot(t, x_sum, label='Somme')
plt.plot(t, x_env, '--', label='Enveloppe')
plt.plot(t, -x_env, '--', label='-Enveloppe')
plt.xlabel('Temps (s)')
plt.legend()
plt.grid()
plt.show()

# Explication : la somme de deux sinusoïdes de fréquences proches donne une oscillation rapide (fm) modulée par une enveloppe lente (fd).
```

### Exercice 9 – Repliement de spectre (aliasing) – approche numérique
Échantillonner le signal $x(t) = \cos(2\pi 30 t)$ avec une fréquence d’échantillonnage $f_e = 20$ Hz (donc sous-échantillonnage).  
Générer les échantillons pour $t = 0, 0.05, 0.10, \dots$ et les tracer.  
Reconstruire le signal à partir des échantillons en traçant la droite reliant les points (ou en utilisant une interpolation).  
Quelle fréquence apparente observez-vous ? Justifier à l’aide de la formule d’aliasing.



**Code** :

$f=30$ Hz, $f_e=20$ Hz $<2f$ : condition de Shannon non respectée.
Formule de repliement : $f_{app}=\left|f-k f_e\right|$ avec $k$ entier minimisant le résultat, ici $|30-20|=10$ Hz.
Vérification : $\cos(2\pi\cdot30\cdot n/20)=\cos(3\pi n)=(-1)^n=\cos(2\pi\cdot10\cdot n/20)$ : les échantillons sont **identiques** à ceux d'un cosinus de 10 Hz.
La fréquence apparente est donc **10 Hz** (la reconstruction linéaire donne un signal en dents de scie alternant $\pm1$, cas limite : 10 Hz = $f_e/2$).

```python
fe = 20
f = 30
t_cont = np.linspace(0, 0.5, 1000)
x_cont = np.cos(2*np.pi*f*t_cont)



# Échantillonnage
te = 1/fe
t_ech = np.arange(0, 0.5, te)
x_ech = np.cos(2*np.pi*f*t_ech)

plt.plot(t_ech, x_ech, 'g--', label='Reconstruction linéaire')
plt.plot(t_cont, x_cont, label='Signal original (30 Hz)')
plt.stem(t_ech, x_ech, linefmt='r-', markerfmt='ro', basefmt=' ', label='Échantillons (20 Hz)')

# Reconstruction linéaire (simple interpolation)
t_recon = np.linspace(0, 0.5, 1000)
x_recon = np.interp(t_cont, t_ech, x_ech)
plt.plot(t_cont, x_recon, 'g--', label='Reconstruction linéaire')

plt.legend()
plt.grid()
plt.show()

# La fréquence apparente est |fe - f| = 10 Hz. En effet, l'échantillonnage à 20 Hz donne un alias à 10 Hz.
```

### Exercice 10 – Synthèse d’un signal carré par séries de Fourier
Un signal carré de période $T$ et d’amplitude 1 peut être décomposé en série de Fourier :  
$$
x(t) = \frac{4}{\pi} \sum_{k=0}^{N} \frac{1}{2k+1} \sin\bigl(2\pi (2k+1) f_0 t\bigr).
$$  
Écrire une fonction Python qui génère ce signal pour un nombre d’harmoniques $N$ donné.  
Tester pour $N=1, 5, 20, 50$ et tracer les signaux correspondants.  
Observer le phénomène de Gibbs (oscillations près des discontinuités).  
Comparer avec la fonction `square` de `scipy.signal`.

**Code** :

```python
%pip install scipy
```

```python
import scipy
from scipy import signal

print("Version de SciPy :", scipy.__version__)
print("SciPy est installé correctement !")
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal

def carre_fourier(t, f0, N):
    x = np.zeros_like(t, dtype=float)

    for k in range(N):
        n = 2 * k + 1
        x += np.sin(2 * np.pi * n * f0 * t) / n

    return (4 / np.pi) * x
```

```python
def carre_fourier(t, f0, N):
    '''Signal carré (amplitude 1) par N harmoniques impaires : (4/π) Σ sin(2π(2k+1)f0 t)/(2k+1).'''
    x = np.zeros_like(t)
    for k in range(N):
        n = 2*k + 1
        x += np.sin(2*np.pi*n*f0*t)/n
    return 4/np.pi*x

f0 = 1
t = np.linspace(0, 2, 4000, endpoint=False)
fig, axes = plt.subplots(2, 2, figsize=(12, 7))
for ax, N in zip(axes.flat, [1, 5, 20, 50]):
    ax.plot(t, carre_fourier(t, f0, N), label=f'N={N}')
    ax.axhline(1, color='gray', lw=.5); ax.grid(); ax.legend()
plt.suptitle("Synthèse d'un signal carré par séries de Fourier"); plt.tight_layout(); plt.show()

plt.plot(t, signal.square(2*np.pi*f0*t), label='scipy.signal.square')
plt.plot(t, carre_fourier(t, f0, 50), '--', label='N=50'); plt.legend(); plt.grid(); plt.show()
t_fin = np.linspace(0, 1, 200000, endpoint=False)   # grille fine pour bien capter le pic
for N in (5, 20, 50, 200):
    mx = carre_fourier(t_fin, f0, N).max()
    print(f"N={N:3d} : max = {mx:.4f} (dépassement ≈ {100*(mx-1)/2:.1f} % du saut)")
```

### Exercice 11 – Phase et décalage temporel
Montrer que $A\cos(2\pi f t - \phi)$ est équivalent à un décalage temporel. Quel est le décalage en secondes ?  
Pour $A=1, f=2$ Hz, $\phi = \pi/2$, calculer le décalage. Tracer le signal non décalé et le signal décalé pour vérifier.

**Démonstration** : $\cos(2\pi f (t - \Delta t)) = \cos(2\pi f t - 2\pi f \Delta t)$ donc $\phi = 2\pi f \Delta t$ et $\Delta t = \phi/(2\pi f)$.

**Code** :

$\cos(2\pi f(t-\Delta t))=\cos(2\pi ft-2\pi f\Delta t)$, donc $\phi=2\pi f\Delta t$ et $\Delta t=\dfrac{\phi}{2\pi f}$ :
un déphasage $\phi$ est un retard de $\Delta t$.
Pour $f=2$ Hz, $\phi=\pi/2$ : $\Delta t=\dfrac{\pi/2}{4\pi}=0{,}125$ s (soit $T/4$ avec $T=0{,}5$ s).

```python
f = 2
phi = np.pi/2
A = 1

delta_t = phi / (2*np.pi*f)
print(f"Décalage temporel : {delta_t:.4f} s")

t = np.linspace(0, 1, 1000)

x_original = A * np.cos(2*np.pi*f*t)
x_decale = A * np.cos(2*np.pi*f*(t - delta_t))
x_phase = A * np.cos(2*np.pi*f*t - phi)

plt.plot(t, x_original, label='Original (φ=0)')
plt.plot(t, x_decale, '--', label='Décalé de Δt')
plt.plot(t, x_phase, ':', lw=3, label='Phase φ = π/2')

plt.legend()
plt.grid()
plt.xlabel('Temps (s)')
plt.show()

print("Décalé = déphasé :", np.allclose(x_decale, x_phase))
```

### Exercice 12 – Échantillonnage d’une exponentielle complexe
Soit $z[n] = e^{j\omega_0 n}$. Montrer que si $\omega_0$ est un multiple rationnel de $2\pi$, le signal est périodique.  
Pour $\omega_0 = 2\pi/3$, déterminer la période fondamentale.  
Pour $\omega_0 = 1$ rad/échantillon, le signal est-il périodique ? Justifier.  
Générer les deux signaux sur 50 échantillons et observer leur comportement.
**Démonstration** : La périodicité exige $\omega_0 N = 2\pi k$ avec $k$ entier, donc $\omega_0/(2\pi) = k/N$ est rationnel.
**Code** :

$z[n]=e^{j\omega_0n}$ est périodique de période $N$ ssi $z[n+N]=z[n]$, i.e. $\omega_0N=2\pi k$ ($k\in\mathbb Z$), soit $\dfrac{\omega_0}{2\pi}=\dfrac kN\in\mathbb Q$.
- $\omega_0=2\pi/3$ : $\omega_0/2\pi=1/3$, **période fondamentale $N=3$**.
- $\omega_0=1$ rad/éch. : $\omega_0/2\pi=1/(2\pi)$ est irrationnel, donc **non périodique** (le motif ne se répète jamais exactement, même si le signal reste « oscillant »).


```python
import numpy as np
import matplotlib.pyplot as plt

n = np.arange(50)

# Signal 1 : omega0 = 2*pi/3
z1 = np.exp(1j * (2*np.pi/3) * n)

# Signal 2 : omega0 = 1 radian/échantillon
z2 = np.exp(1j * 1 * n)

# Affichage des deux signaux
fig, ax = plt.subplots(1, 2, figsize=(14, 4))

ax[0].stem(n, z1.real, basefmt=' ')
ax[0].set_title(r'$\omega_0 = 2\pi/3$ (périodique, N = 3)')

ax[1].stem(n, z2.real, basefmt=' ')
ax[1].set_title(r'$\omega_0 = 1$ (non périodique)')

for a in ax:
    a.grid()
    a.set_xlabel('n')
    a.set_ylabel('Partie réelle')

plt.tight_layout()
plt.show()

# Vérification de la périodicité
print("z1 périodique (N=3) :", np.allclose(z1[3:], z1[:-3]))

print(
    "z2 : plus petite période N <= 40 :",
    next(
        (N for N in range(1, 41)
         if np.allclose(z2[N:], z2[:-N])),
        None
    )
)
```
